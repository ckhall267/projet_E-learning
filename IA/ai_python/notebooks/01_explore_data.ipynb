# Exploration des données

Ce notebook explore le dataset des symptômes et maladies.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add parent directory to path
sys.path.append(os.path.dirname(os.path.dirname(os.path.abspath(''))))

from src.data_preparation import DataPreparator

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## Chargement des données


```python
# Initialize preparator
preparator = DataPreparator(data_path="../data/raw/symptoms_disease.csv")

# Load data
df = preparator.load_data()
df.head()
```

## Analyse des données


```python
# Basic info
print(f"Shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
print(f"\nMissing values:\n{df.isnull().sum()}")
print(f"\nData types:\n{df.dtypes}")
```

```python
# Clean data
df_clean = preparator.clean_data(df)
df_clean.head()
```

## Distribution des maladies


```python
# Find disease column
disease_col = [col for col in df_clean.columns if 'disease' in col.lower()][0]

# Plot distribution
disease_counts = df_clean[disease_col].value_counts()

plt.figure(figsize=(14, 6))
disease_counts.plot(kind='bar')
plt.title('Distribution des maladies')
plt.xlabel('Maladie')
plt.ylabel('Nombre de cas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print(f"Nombre de maladies uniques: {df_clean[disease_col].nunique()}")
print(f"\nDistribution:\n{disease_counts}")
```

## Analyse des symptômes


```python
# Get symptom columns
symptom_cols = [col for col in df_clean.columns if col != disease_col]

print(f"Nombre de symptômes: {len(symptom_cols)}")

# Most common symptoms
symptom_freq = df_clean[symptom_cols].sum().sort_values(ascending=False)

plt.figure(figsize=(14, 6))
symptom_freq.head(20).plot(kind='bar')
plt.title('Top 20 symptômes les plus fréquents')
plt.xlabel('Symptôme')
plt.ylabel('Fréquence')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Corrélations


```python
# Correlation matrix for top symptoms
top_symptoms = symptom_freq.head(15).index.tolist()
corr_matrix = df_clean[top_symptoms].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Matrice de corrélation des symptômes')
plt.tight_layout()
plt.show()
```
